### Useful Functitons

```python
'''
for adding another column 'Stars' based on the IMBD_Rating
>9, *****
>8, ****
otherwise, ***
'''
```

```python
# Custom function creation - example
# Function to assign stars based on the IMDB_Rating
def starts_fn(rating):
    if rating > 9:
        return "*****"
    elif rating > 8:
        return "****"
    else: return "***" 
```

```python
#Command for using apply function
df['Stars'] = df['IMDB_Rating'].apply(starts_fn)
df
```
