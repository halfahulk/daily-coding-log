### Merging Dataframes

```python
Groups_1 = {'Gr_SN': [1, 2, 3, 4], 'Group_Name': ['X','Lions','Y','ABC']}
Groups_2 = {'Gr_ID': [12, 1, 3, 15], 'Group_Name': ['Lions', 'Dev', 'Y', 'ZZZ']}
```

```python
G_df_1 = pd.DataFrame(Groups_1)
G_df_2 = pd.DataFrame(Groups_2)
```

```python
G_df_1
```

```python
G_df_2
```

```python
#Command to merge one dataframe to another

#Inner Merge

pd.merge(G_df_1, G_df_2, how='inner', on='Group_Name')
```

```python
#Left Merge

pd.merge(G_df_1, G_df_2, how='left')
```
